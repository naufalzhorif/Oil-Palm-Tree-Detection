## Training Result

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Read csv

```python
df = pd.read_csv("checkpoints/run_280926_1013/training_log.csv")
```

```python
df.head()
```

```python
plt.rcParams.update({
    "font.family":        "serif",
    "font.serif":         ["Times New Roman", "DejaVu Serif"],
})

C_TRAIN = "#0072B2" 
C_VAL   = "#D55E00"
C_MAP05 = "#0072B2" 
C_MAP75 = "#D55E00"
C_MAP95 = "#009E73"   
C_GRID  = "#D9D9D9"
```

### Plotting

```python
epochs = df["epoch"]

precision     = df["precision"]
val_precision = df["val_precision"]
recall        = df["recall"]
val_recall    = df["val_recall"]

f1     = df["f1"]
val_f1 = df["val_f1"]
mAP05  = df["mAP_0.5"]
mAP075 = df["mAP_0.75"]
mAP0595 = df["mAP_0.5_0.95"]


fig, axes = plt.subplots(2, 2, figsize=(11, 7), sharex=True)
fig.suptitle("Faster R-CNN Training Result", fontsize=14, fontweight="bold")

# Precision
ax = axes[0, 0]
ax.plot(epochs, precision,     label="Training",   color=C_TRAIN)
ax.plot(epochs, val_precision, label="Validation", color=C_VAL)
ax.set_ylabel("Precision", fontsize=12)
ax.legend(fontsize=9, loc="lower right")
ax.grid(True, linestyle="--", alpha=0.5)

# Recall
ax = axes[0, 1]
ax.plot(epochs, recall,     label="Training",   color=C_TRAIN)
ax.plot(epochs, val_recall, label="Validation", color=C_VAL)
ax.set_ylabel("Recall", fontsize=12)
ax.legend(fontsize=9, loc="lower right")
ax.grid(True, linestyle="--", alpha=0.5)

# F1-Score
ax = axes[1, 0]
ax.plot(epochs, f1,     label="Training",   color=C_TRAIN)
ax.plot(epochs, val_f1, label="Validation", color=C_VAL)
ax.set_xlabel("Epoch", fontsize=12)
ax.set_ylabel("F1-Score", fontsize=12)
ax.legend(fontsize=9, loc="lower right")
ax.grid(True, linestyle="--", alpha=0.5)

# mAP
ax = axes[1, 1]
ax.plot(epochs, mAP05,   label="mAP@0.5",      color=C_MAP05)
ax.plot(epochs, mAP075,  label="mAP@0.75",     color=C_MAP75)
ax.plot(epochs, mAP0595, label="mAP@0.5:0.95", color=C_MAP95)
ax.set_xlabel("Epoch", fontsize=12)
ax.set_ylabel("mAP", fontsize=12)
ax.legend(fontsize=8, loc="lower right", ncol=3)
ax.grid(True, linestyle="--", alpha=0.5)


plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
